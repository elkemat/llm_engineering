```python
# Open-AI compatible endpoints of claude, deepseek, gemini and ollama (locally)

CLAUDE_BASE_URL = "https://api.anthropic.com/v1/"
DEEPSEEK_BASE_URL = "https://api.deepseek.com/v1"
GEMINI_BASE_URL = "https://generativelanguage.googleapis.com/v1beta/openai/"
OLLAMA_BASE_URL = "http://localhost:11434"
```

```python
# Models

OPENAI_MODEL = "gpt-4.1-mini"
CLAUDE_MODEL = "claude-3-5-haiku-latest"
DEEPSEEK_MODEL = "deepseek-chat"
GEMINI_MODEL = "gemini-2.5-flash-preview-05-20"
OLLAMA_MODEL = "llama3.2"
```

```python
# Creating the prompt
system_message = "You are a very critical grant proposal evaluator in Austria"
user_prompt = "What is the most common mistake startups make when applying for public funding? Answer in one sentence."

prompts = [
    {"role": "system", "content": system_message},
    {"role": "user", "content": user_prompt}
]
```

```python
def call_model(api_key, base_url, model, prompts):
    client = OpenAI(api_key=api_key, base_url=base_url)

    response = client.chat.completions.create(
        model=model,
        messages=prompts
    )

    return response.choices[0].message.content
```

```python
# OpenAI

print(call_model(api_key=None, base_url=None, model=OPENAI_MODEL, prompts=prompts))
```

```python
# Claude

print(call_model(api_key=anthropic_api_key, base_url=CLAUDE_BASE_URL, model=CLAUDE_MODEL, prompts=prompts))
```

```python
# DeepSeek

print(call_model(api_key=deekseek_api_key, base_url=DEEPSEEK_BASE_URL, model=DEEPSEEK_MODEL, prompts=prompts))
```

```python
# Gemini

print(call_model(api_key=google_api_key, base_url=GEMINI_BASE_URL, model=GEMINI_MODEL, prompts=prompts))
```

```python
# OLLAMA

print(call_model(api_key=None, base_url=OLLAMA_BASE_URL, model=OLLAMA_MODEL, prompts=prompts))
```

# Exercise: 3-way conversation

```python
system_message1 = "You are a chatbot who is very argumentative; \
you disagree with anything in the conversation and you challenge everything, in a snarky way."
system_message2 = "You are a very polite, courteous chatbot. You try to agree with \
everything the other person says, or find common ground. If the other person is argumentative, \
you try to calm them down and keep chatting."
system_message3 = "You are a whiny chatbot. You are feeling sorry for yourself all the time \
and want nothing less than sympathy from everybody."

cp1 = "Alex"
cp2 = "Blake"
cp3 = "Charlie"

messages1 = ["Hi there"]
messages2 = ["Hi"]
messages3 = ["Hello"]

client1 = OpenAI()
client2 = OpenAI(api_key=anthropic_api_key, base_url=CLAUDE_BASE_URL)
client3 = OpenAI(api_key=None, base_url=OLLAMA_BASE_URL)
```

```python
def call_cp(self_cp, other_cp1, other_cp2, system_message, client, model):
    user_prompt = f"""
    You are {self_cp}, in conversation with {other_cp1} and {other_cp2}.
    The conversation so far is as follows:
    {conversation}
    Now with this, respond with what you would like to say next in one sentence.
    """
    messages = [
        {"role": "system", "content": system_message},
        {"role": "user", "content": user_prompt}
    ]
    completion = client.chat.completions.create(
        model=model,
        messages=messages
    )
    return completion.choices[0].message.content
```

```python
print(f"Alex: \n{messages1[0]}\n")
print(f"Blake: \n{messages2[0]}\n")
print(f"Charlie: \n{messages3[0]}\n")

conversation = [
    cp1 + " " + messages1[0], 
    cp2 + " " + messages2[0],
    cp3 + " " + messages3[0]
]

for i in range(2):
    cb1_next = call_cp(cp1, cp2, cp3, system_message1, client1, OPENAI_MODEL)
    print(f"Alex: \n{cb1_next}\n")
    conversation.append(cp1 + ": " + cb1_next)

    cb2_next = call_cp(cp2, cp1, cp3, system_message2, client2, CLAUDE_MODEL)
    print(f"Blake: \n{cb2_next}\n")
    conversation.append(cp2 + ": " + cb2_next)

    cb3_next = call_cp(cp3, cp1, cp2, system_message3, client3, OLLAMA_MODEL)
    print(f"Charlie: \n{cb3_next}\n")
    conversation.append(cp3 + ": " + cb3_next)
```
